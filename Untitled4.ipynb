```python
import csv
import os
from datetime import datetime

# --- Constants ---
CSV_FILE = 'expenses.csv'
CSV_HEADERS = ['Date', 'Category', 'Amount', 'Note']

def initialize_csv():
    """Initializes the CSV file with headers if it doesn't exist."""
    if not os.path.exists(CSV_FILE):
        with open(CSV_FILE, 'w', newline='') as f:
            writer = csv.writer(f)
            writer.writerow(CSV_HEADERS)
        print(f"Created new expense file: {CSV_FILE}")

initialize_csv()
```

```python
def load_expenses():
    """Loads all expense records from the CSV file."""
    expenses = []
    if os.path.exists(CSV_FILE):
        with open(CSV_FILE, 'r', newline='') as f:
            reader = csv.DictReader(f)
            for row in reader:
                # Convert amount to float, handle potential errors
                try:
                    row['Amount'] = float(row['Amount'])
                    expenses.append(row)
                except ValueError:
                    print(f"Warning: Skipping row with invalid amount: {row}")
    return expenses

def save_expense(expense_data):
    """Appends a new expense record to the CSV file."""
    with open(CSV_FILE, 'a', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(expense_data)
    print("Expense added successfully!")
```

```python
def add_expense():
    """Prompts the user for expense details and saves the record."""
    date_str = input("Enter date (YYYY-MM-DD, leave blank for today): ")
    if not date_str:
        date = datetime.now().strftime('%Y-%m-%d')
    else:
        try:
            datetime.strptime(date_str, '%Y-%m-%d')
            date = date_str
        except ValueError:
            print("Invalid date format. Using today's date.")
            date = datetime.now().strftime('%Y-%m-%d')

    category = input("Enter category: ").strip()
    while not category:
        print("Category cannot be empty.")
        category = input("Enter category: ").strip()

    while True:
        try:
            amount = float(input("Enter amount: "))
            if amount <= 0:
                raise ValueError
            break
        except ValueError:
            print("Invalid amount. Please enter a positive number.")

    note = input("Enter note (optional): ").strip()

    expense_data = [date, category, amount, note]
    save_expense(expense_data)
```

```python
def view_expenses():
    """Loads and displays all recorded expenses with a total amount."""
    expenses = load_expenses()
    if not expenses:
        print("No expenses recorded yet.")
        return

    print("\n--- All Expenses ---")
    total_amount = 0.0
    # Print headers
    print(f"{CSV_HEADERS[0]:<12} {CSV_HEADERS[1]:<15} {CSV_HEADERS[2]:<10} {CSV_HEADERS[3]:<25}")
    print("-" * 65)

    for expense in expenses:
        print(f"{expense['Date']:<12} {expense['Category']:<15} {expense['Amount']:<10.2f} {expense['Note']:<25}")
        total_amount += expense['Amount']

    print("-" * 65)
    print(f"Total spent: {total_amount:.2f}\n")
```

```python
def category_summary():
    """Displays a summary of spending by category."""
    expenses = load_expenses()
    if not expenses:
        print("No expenses recorded yet.")
        return

    summary = {}
    for expense in expenses:
        category = expense['Category'].lower()
        summary[category] = summary.get(category, 0.0) + expense['Amount']

    print("\n--- Category-wise Spending Summary ---")
    if summary:
        for category, amount in sorted(summary.items()):
            print(f"{category.capitalize()}: {amount:.2f}")
    else:
        print("No spending categories found.")
    print("\n")
```

```python
def main_menu():
    """Displays the main menu and handles user choices."""
    while True:
        print("--- Expense Tracker Menu ---")
        print("1. Add New Expense")
        print("2. View All Expenses")
        print("3. View Category Summary")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ").strip()

        if choice == '1':
            add_expense()
        elif choice == '2':
            view_expenses()
        elif choice == '3':
            category_summary()
        elif choice == '4':
            print("Exiting Expense Tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")
        input("Press Enter to continue...") # Pause for user to read output
```
